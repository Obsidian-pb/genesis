```python
import osmnx as ox
import geopandas as gpd
```

```python
polygons_gdf = gpd.read_file('_data/kemerovskaya_obl.gpkg', layer = '__')
polygons_gdf
```

```python
if polygons_gdf.crs != 'EPSG:4326':
    polygons_gdf = ox.projection.project_gdf(polygons_gdf, to_latlong = True)
```

```python
SETTLEMENT_TAGS = {
    'place': ['city'],
    # 'place': ['city', 'town', 'village', 'hamlet'],
}
settlements = ox.features_from_polygon(polygons_gdf.union_all(), SETTLEMENT_TAGS)
settlements
```

```python
settlements.query('place == "city"')
```

```python
settlements2 = settlements.copy()
settlements2.reset_index(inplace=True)
if 'element' in settlements2.columns:
    settlements2.drop('element', axis=1, inplace=True)
if 'id' in settlements2.columns:
    settlements2.drop_duplicates('id', inplace=True)
    settlements2.set_index('id', inplace=True)  #, verify_integrity = True)
    

boundaries = settlements2[settlements2.geometry.apply(lambda x: x.geom_type).isin(['Polygon', 'MultiPolygon'])]
boundaries.query('place == "city"')
```
