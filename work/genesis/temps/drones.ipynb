# Тесты расчетов для дронов

```python
import geopandas as gpd
from shapely import Point
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


```

```python
def get_points_env(n,m, poligon, step=100):
    points = []
    for i in range(n):
        for j in range(m):
            pnt = Point(i*step,j*step)
            points.append(pnt)

    return points

points_env = get_points_env(100,100, poligon=None)
gpd.GeoDataFrame(geometry=points_env).plot()
```

```python
def get_distance(pnt):
    return pnt.distance(Point(1000, 3000))

env = gpd.GeoDataFrame(geometry=points_env)
env['distance'] = env.geometry.apply(get_distance)
env.head(3)
```

```python
env.plot(column='distance', cmap='Set1', legend=True)
```

```python
env.plot(column='distance', cmap='Grays', legend=True)
```

```python
ax = env.plot(cmap='Grays', markersize=5, marker='s')
env.query('distance < 3000').plot(ax=ax, color='r', markersize=5, marker='s')
```
