```python
import osmnx as ox
```

```python
G = ox.graph_from_place('Железногорск, Красноярский край, Россия', network_type='drive_service')

edges = ox.graph_to_gdfs(G, nodes=False)
```

```python
for col in edges:
    print(col, end=': ')
    print(edges[col].dtype)
```

```python
edges = ox.graph_to_gdfs(G, nodes = False)
for id, data in edges[:5].iterrows():
    print(data.geometry)
```
